# Python Functions

```python
def greet():
    print("Good Morning")
greet()
```

```python
def addition(a,b):
    c = a + b
    return c
addition(12,8)
```

```python
def subs(a,b):
    c = a - b
    return c
subs(12,8)
```

```python
def add_sub(x,y):
    c = x + y
    d = x - y
    e = x * y
    return c,d,e
# function will return tuple by default
add_sub(12,8)
```

```python
# converting tuple to seperate variables
add, sub, mul = add_sub(12,8)
print(f"12 + 8 : {add}\n12 - 8 : {sub}\n12 * 8 : {mul}")
```
