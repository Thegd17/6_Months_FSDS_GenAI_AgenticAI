```python
#1
a = input("Is A is angry? ").lower()
b = input("Is B is angry? ").lower()

if a == 'yes' and b == 'yes':
    print("Both are angry")
elif a == 'no' and b == 'no':
    print("None of them is angry")
else:
    print("One of them is angry")
```
